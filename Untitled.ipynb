```python
from IPython.display import clear_output
mapa = [["0","0","0"],
         ["0","0","0"],
         ["0","0","0"]]
jugador_fila = 1
jugador_columna = 1
while True:
  for i in range(3):
    for j in range(3):
      mapa[i][j] = "0"
      mapa[jugador_fila][jugador_columna] = "p"
      for fila in mapa:
        print(fila)
      clear_output(wait=True)
      movimiento = input("mover w,a,s,d o salir")
      mapa[jugador_fila][jugador_columna] = "0"
      if movimiento.lower() == "salir":
        break
      elif movimiento.lower() == "w" and jugador_fila > 0:
        jugador_fila -= 1
      elif movimiento.lower() == "s" and jugador_fila < 2:
        jugador_fila += 1
      elif movimiento.lower() == "d" and jugador_columna < 2:
        jugador_columna += 1
      elif movimiento.lower() == "a" and jugador_columna > 0:
        jugador_columna -= 1
print("ADIOS")





```
